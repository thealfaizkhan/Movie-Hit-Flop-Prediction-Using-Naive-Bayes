## Movie Hit/Flop Prediction Using Naive Bayes

```python
# libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import confusion_matrix,accuracy_score
```

```python
# Dataset Load
data= pd.read_csv(r"C:\Users\ALFAIZ KHAN\Downloads\movie_hit_flop_dataset.csv")
print(data.head(10))
```

```python
# Chexk a dataset columns indexing and datatype
data.columns
```

```python
#Check Null Value in dataset 
data.isnull().sum()
```

```python
# Label Encoder use se text convert to Numbers
encoder = LabelEncoder()
data ['label']=encoder.fit_transform(data.label)
```

```python
encoder
```

```python
# First column Index with All rows with store x variable
x = data.iloc[:,0]
print(x)
```

```python
# Last Column with all rows store y variable
y = data.iloc[0:,1]
print(y)
```

```python
# Data ko Train and Testing 25% Data 
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.25, random_state = 0 )
```

```python
x_train.shape
```

```python
y_test.shape
```

```python
# Input x ko wor-count numbers me convert ker deta hai
vector = CountVectorizer(stop_words ='english')
vector.fit(x_train)
```

```python
x_train_transformed = vector.transform(x_train)
x_test_transformed = vector.transform(x_test)
```

```python
# Model Train
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(x_train_transformed,y_train)
```

```python
# Prediction
y_pred = model.predict(x_test_transformed)
y_pred_prob = model.predict_proba (x_test_transformed)
y_pred_prob
```

```python
len (y_pred)
```

```python
# Check a accuracy and confusion matrix..
print(confusion_matrix(y_test,y_pred))
print()
print(accuracy_score(y_test,y_pred))
```

## Movie Hit/Flop Prediction Using Naive Bayes

## Project Overview
This project uses Machine Learning to predict whether a movie is a **Hit or Flop** based on its review text. It applies the Multinomial Naive Bayes algorithm to classify movie reviews using Natural Language Processing (NLP) techniques.

## Technologies Used
- Python
- Pandas
- Scikit-learn
- CountVectorizer
- LabelEncoder
- Multinomial Naive Bayes
- Joblib

## Project Workflow
1. Loaded the movie review dataset using Pandas.
2. Checked the dataset columns and missing values.
3. Converted categorical labels (Hit/Flop) into numerical values using LabelEncoder.
4. Split the dataset into training and testing sets using an 75:25 ratio.
5. Converted text reviews into numerical word-count features using CountVectorizer.
6. Trained a Multinomial Naive Bayes classification model.
7. Evaluated the model using accuracy score and a confusion matrix.
8. Built an interactive command-line prediction system to classify new movie reviews.
9. Saved the trained model, CountVectorizer, and LabelEncoder using Joblib for future predictions.

## Results
The model achieved **100% accuracy on the current test split**, with 38 test samples correctly classified. This result applies only to the current dataset and split; further testing on unseen, real-world reviews is needed to assess generalization.

## Conclusion
This project demonstrates a basic end-to-end text classification workflow using NLP and Machine Learning. It can serve as a foundation for sentiment analysis and movie review classification projects.
